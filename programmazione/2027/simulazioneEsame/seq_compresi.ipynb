## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def seq_compresi(sequenza, low, high):

    cnt = 0
    for v in sequenza:
        if low <= v <= high:
            cnt += 1
    return cnt
```

## Traccia

Scrivere una funzione `seq_compresi(L,low,high)` che restituisca il

(estremi inclusi). Potete assumere che gli argomenti siano di tipo
corretto, cioè una lista di numeri e due numeri.

Perciò **non è necessario** che verifichiate queste condizioni o che

```python
seq = [3,5,1,8,4,2,5,7,6,3,4,1,9,8,4]
n = seq_compresi(seq,3,5)
print(n)
```
