## Traccia

Quale sarà il valore della variabile `somma` dopo l'esecuzione del ciclo for

```python
somma=0

L = [1, 1, 2, 2, 7, 7, -3, -3, 6, 6]

for i in range(3,9):
    somma += L[i]
```
