# Maximum element position in a numeric matrix
Restituisce la posizione (riga, colonna) del massimo in una matrice. In caso di più massimi restituisce la posizione minima (secondo l'ordinamento lessicografico sulla coppia (riga, colonna)) di un massimo

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def good_mat(m):
    if not isinstance(m, list):
        return False
    for row in m:
        if not isinstance(row, list):
            return False
    if len(m) == 0:
        return False

    if len(m[0]) == 0:
        return False

    for row in m:
        if len(row) != len(m[0]):
            return False
    return True


def mat_posmax(m):

    if not good_mat(m):
        raise TypeError

    R = len(m)
    C = len(m[0])
    massimo = m[0][0]
    massimo += 0  # to get TypeError if not num
    pos = (0, 0)

    for i in range(R):
        for j in range(C):
            if massimo < m[i][j]:
                massimo = m[i][j]
                pos = (i, j)
    return pos
```

## Traccia



Write a function `mat_posmax(m)` that:

-   takes as argument a matrix `m`;
-   returns a pair of integers, e.g. by a `return i, j`,
    that are the row and column indices where the maximum element in the matrix appears. In case of more maxima, the position of the first occurrence must be returned (minimum row index, and in case of more maxima in that row minimum column index)
