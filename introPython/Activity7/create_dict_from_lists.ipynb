## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python


def create_dict_from_lists(k_list, v_list):

    res = {}
    for i in range(len(k_list)):
        key = k_list[i]
        value = v_list[i]
        res[key] = value
    return res
```

## Traccia



Write a function `create_dict_from_lists(n)` that:

-   takes as argument two lists `k_list` and `v_list` with equal lengths;
-   returns a `dict` whose keys are taken from elements in `k_list` and values are taken from `v_list`. Element `v_list[0]` will be associated to `k_list[0]`, element `v_list[1]` will be associated to `k_list[1]` and so on.

For example:

```python
print(create_dict_from_lists(["cat", "dog", "mouse"],
                             ["gatto", "cane", "topo"]))
```
