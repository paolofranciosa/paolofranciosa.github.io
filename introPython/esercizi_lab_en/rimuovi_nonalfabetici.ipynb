## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

````python
def rimuovi_nonalfabetici(frase):

    if not isinstance(frase, str):
        raise TypeError

    text = []

    for c in frase:
        if not c.isalpha():
            text.append(' ')
        else:
            text.append(c)
    return "".join(text)
````

Notare la presenza di un elemento `" "` in corrispondenza dei caratteri

di nome `pippo`, è una lettera si può usare il metodo `pippo.isalpha()`,
