## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python


def crea_mat(nrighe, ncol, val=0):
    if nrighe < 1 or ncol < 1:
        raise ValueError

    M = []
    for i in range(nrighe):
        M.append([val] * ncol)
    return M

def verifica_mat(m):
    if not isinstance(m, list):
        return False
    for row in m:
        if not isinstance(row, list):
            return False
    if len(m) == 0:
        return False

    if len(m[0]) == 0:
        return False

    for row in m:
        if len(row) != len(m[0]):
            return False
    return True

def trasposta_mat(A):
    if not verifica_mat(A):
        raise TypeError
    r, c = len(A), len(A[0])
    T = crea_mat(c, r)
    for i in range(c):
        for j in range(r):
            T[i][j] = A[j][i]
    return T
```

## Traccia



Scrivere una funzione `trasposta_mat(m)` che:

-   si aspetti come argomento matrice `m`,
-   crei e restituisca la matrice trasposta di `m`. 

Qui è utile prima creare la matrice contenente elementi qualsiasi, ad esempio 0, come visto a lezione, per poi modificarne gli elementi in modo da avere la trasposta di `m`.

Ad esempio, se viene passata come parametro la matrice `[[2,  3, 7], [5, 3,  1]]` deve essere restituita la matrice `[[2, 5],  [3, 3], [7, 1]]`.
  
Ricordiamo che in Python noi rappresentiamo una matrice $r \times c$ come
una lista di lunghezza $r$, contenente $r$ liste tutte quante di
lunghezza $c$. Potete sempre assumere che l\'input di questo esercizio
sia una matrice ben formata, con $r \geq  1$ e $c \geq 1$, e che tutte
le liste che rappresentano le righe contengano $c$ valori numerici.
