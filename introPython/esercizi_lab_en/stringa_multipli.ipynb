## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def stringa_multipli(n, s):
    i = 1
    res = ''
    while (i < n) :
        res = res + str(s*i) + ' '
        i = i + 1
    res = res + str(s*i) + '.'
    return res
```

## Traccia

Esempi:
