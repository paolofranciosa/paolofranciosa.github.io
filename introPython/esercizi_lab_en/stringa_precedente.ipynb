## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def stringa_precedente(s1, s2):
    if s1 < s2:
        return s1
    else:
        return s2
```

## Traccia

Scrivete una funzione `stringa_precedente(s1, s2)` che restituisca la stringa tra `s1` e `s2` che viene prima in ordine alfabetico. Ad esempio

```python
print(stringa_precedente('Zucca', 'ananas'))
print(stringa_precedente('zucca', 'ananas'))
print(stringa_precedente('20', '100'))
```
