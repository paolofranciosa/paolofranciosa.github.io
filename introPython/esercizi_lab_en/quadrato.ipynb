## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def quadrato(n):
    righe = 0
    res = ''
    while righe < n:
        colonne = 0
        while colonne < n:
            res = res + '*'
            colonne = colonne + 1
        res = res + '\n'
        righe = righe + 1
    return res
```

## Traccia 

Scrivere una funzione `quadrato(x)` che:

-   si aspetti come argomento un numero non-negativo `x`;
