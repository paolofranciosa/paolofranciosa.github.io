## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def tronca_lista_pari_ogni_due(L):
    if len(L) < 2:
        return None
    
    i_trunc = len(L)
    for i in range(0, len(L)-1):
        if L[i] % 2 != 0 and L[i+1] % 2 != 0:
            i_trunc = i
            break        
            
    for i in range(1,len(L)-i_trunc):
        L.pop()
```

## Traccia 


Scrivere una funzione `tronca_lista_pari_ogni_due(L)` che:

- si aspetti come argomento una lista di numeri `L`;
- modifichi direttamente la lista `L` lasciando solo il prefisso più lungo possibile in cui si verifica la condizione che ci sia almeno un numero pari ogni due elementi.

Se la lista contiene meno di due elementi, non va modificata. 

Per questo esercizio si può usare il metodo `.pop()` che rimuove l'ultimo elemento della lista.

#### Esempio

- Se `L = [4, 6, 3, 7, 8]`, la funzione dovrebbe modificare `L` a `[4, 6, 3]`, perché aggiungenzo 7 ci sarebbero due dispari consecutivi.
- Se `L = [4, 6, 3, 2, 1]`, la funzione dovrebbe lasciare `L` invariata a `[4, 6, 3, 2, 1]`.

- Se `L = [3, 3, 3, 7, 8]`, la funzione dovrebbe modificare `L` a `[3]`.

```python
L1 = [4, 6, 3, 7, 8]
tronca_lista_pari_ogni_due(L1)
print(L1)

L2 = [4, 6, 3, 2, 1]
tronca_lista_pari_ogni_due(L2)
print(L2)

L3 = [3, 3, 3, 7, 8]
tronca_lista_pari_ogni_due(L3)
print(L3)
```
