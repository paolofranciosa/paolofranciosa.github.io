## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def medie_mobili(seq, k):
    if not isinstance(seq, list):
        raise TypeError
    for x in seq:
        if not isinstance(x, (int, float)):
            raise TypeError

    if len(seq) < k or k < 1:
        raise ValueError

    medie = [0]
    for i in range(k):
        medie[0] += seq[i]

    for i in range(k, len(seq)):
        medie.append(medie[-1] + seq[i] - seq[i - k])
    for i in range(len(medie)):
        medie[i] /= k

    return medie


def posmax_medie_mobili(seq, k):
    mm = medie_mobili(seq, k)

    pos, maxmm = 0, mm[0]

    for i, x in enumerate(mm):
        if x > maxmm:
            pos, maxmm = i, x
    return pos
```
