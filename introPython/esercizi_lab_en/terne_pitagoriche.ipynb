## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def terne_pitagoriche(N):
    count = 0
    c = 1
    while c < N+1:
        b = 1
        while b < c:
            a = 1
            while a < b:
                if a**2 + b**2 == c**2:
                    #print(a,b,c)
                    count = count + 1
                a += 1
            b += 1
        c += 1
                    
    return count 
```

## Traccia 

Scrivere una funzione `terne_pitagoriche(N)` che:

-   si aspetti come argomento un numero intero `N`;
