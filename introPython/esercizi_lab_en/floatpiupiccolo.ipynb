## Traccia

I numeri `float` sono rappresentazioni dei numeri reali che hanno
una precisione limitata e un dominio limitato, a causa del fatto che ogni float occupa una quantità di memoria fissa (al contrario degli `int` che possono crescere a piacere e quindi occupare una quantità di memoria potenzialmente illimitata).

Entrate nell'interprete interattivo e cercate di capire quale sia il
numero `float` più piccolo che possa essere rappresentato tra
quelli **maggiori di zero**, e che però non sia approssimato a zero da
python.

*Suggerimento:* il modo più rapido per costruire numeri piccoli è quello
di prendere un numero strettamente compreso tra 0 e 1 ed elevarlo a
potenza. Più la potenza cresce più il numero si rimpicciolisce, ma non è
mai zero. Ad un certo punto può darsi però che python non sia più in
grado di rappresentare un numero così piccolo e lo consideri zero.
