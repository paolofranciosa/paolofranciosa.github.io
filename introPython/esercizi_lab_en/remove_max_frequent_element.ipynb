## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def remove_max_frequent_element(a):
    if len(a) == 0:
        return None
    max_f = 0
    for elemento in a:
        f = a.count(elemento)
        if f > max_f:
            ele_m = elemento
            max_f = f
            
    while ele_m in a:
        a.remove(ele_m)
```

## Traccia 

Write a `remove_max_frequent_element(L)` function which:

- expects a list `L` as a parameter,
- directly modifies the list `L` by removing all occurrences of the element that appears with the maximum frequency. If the list is empty, modify nothing.

Assume that there are no equalities, i.e. that the most frequent element is unique in the list.

#### Example

If `L = [3, 5, 3, 7, 8]`, the most frequent element is `3`. After calling `remove_max_frequent_element(L)`, `L` should become `[5, 7, 8]`.


```python
L = [3, 5, 3, 7, 3, 8]
remove_max_frequent_element(L)
print(L)
```
