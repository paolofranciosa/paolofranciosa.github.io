## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def buona_password(stringa):
    hasNumero = False
    for carattere in stringa:
        if carattere.isdigit():
            hasNumero = True

    hasMaiuscola = False
    for carattere in stringa:
        if carattere.isupper():
            hasMaiuscola = True
            
    hasMinuscola = False
    for carattere in stringa:
        if carattere.islower():
            hasMinuscola = True            
        
    hasSpeciale = False
    for carattere in stringa:
        if not carattere.isalpha() and not carattere.isdigit():
            hasSpeciale = True
            
    return hasNumero and hasMaiuscola and hasMinuscola and hasSpeciale
```

## Traccia

Scrivete una funzione `buona_password(s)` che prende una stringa `s` in input e verifica (restituisce True) se la stringa in input contiene almeno un numero, una maiuscola, una minuscola e un simbolo speciale. Ad esempio

```python
print(buona_password('c1aO#'))
print(buona_password('ciaO#'))
print(buona_password('c1ao#'))
print(buona_password('c1aO'))
```

Per questo esercizio potreste usare i metodi di stringa

-   `str.isalpha()` che restituisce `True` quando
    una stringa è **non vuota** ed è fatta di soli caratteri alfabetici;
-   `str.isdigit()` che restituisce `True` quando
    una stringa è **non vuota** ed è fatta di soli cifre numeriche.
-   `str.islower()` che restituisce `True` quando
    una stringa è **non vuota** ed è fatta di soli caratteri minuscoli;
-   `str.isupper()` che restituisce `True` quando
    una stringa è **non vuota** ed è fatta di soli caratteri maiuscoli;

Ad esempio:

```python
print('à'.isalpha())
print('3'.isdigit())
print('a'.islower())
print('A'.isupper())
```
