## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def rimuovi_duplicati_in_place(lista):
    if len(lista) == 0:
        return []

    results = [lista[0]]
    lista_copy = lista[:]

    for parola in lista_copy[1:]:
        if parola != results[-1]:
            results.append(parola)
        else:
            lista.remove(parola)
```

## Traccia

Scrivere una funzione `rimuovi_duplicati_in_place(lista)` che:

- si aspetti come argomento una lista **ordinata** `lista`;
- modifichi direttamente la lista `lista` rimuovendo tutti i duplicati in-place, mantenendo un solo esemplare di ciascun elemento.

Per questo esercizio non serve controllare che il parametro ricevuto sia effettivamente una lista ordinata.

#### Esempio

Se `lista = ["buongiorno", "dall", "e", "e", "e", "in", "torno", "umbria", "umbria", "vengo"]`, dopo aver chiamato `rimuovi_duplicati_in_place(lista)`, `lista` dovrebbe diventare `["buongiorno", "dall", "e", "in", "torno", "umbria", "vengo"]`.

```python
lista = ["buongiorno", "dall", "e", "e", "e", "in", "torno", "umbria", "umbria", "vengo"]
rimuovi_duplicati_in_place(lista)
print(lista)
```
