## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_azzera(d, lista_chiavi):
    if not isinstance(d, dict):
        raise TypeError
    for k in d:
        if k in lista_chiavi:
            d[k] = 0
    return
 
```

## Traccia


Scrivete una funzione `dict_azzera(d, lista_chiavi)` che:

-   si aspetti come argomenti un dizionario e una lista di chiavi;
-   modifichi il dizionario azzerando i valori associati alle chiavi del dizionario che compaiono in lista_chiavi. 

Ad esempio:

```python
diz = {'dog': 3, 'cat': 9, 'mouse': 2, 'fish': -3}
lista = ['dog', 'mouse']
dict_azzera(diz, lista)
print(diz)
```
