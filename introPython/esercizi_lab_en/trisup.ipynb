## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione.

```python
def trisup(M):
    for r in range(0, len(M)):
        for c in range(0, len(M[0])):
            if r > c and M[r][c] != 0:
                return False
    return True  
```

## Traccia


Scrivere una funzione `trisup(M)` che,

-   si aspetti come argomento una matrice *QUADRATA* `M` ben formata;
-   restituisca `True` se la matrice è simmetrica e `False` altrimenti.

Si ricorda che una matrice è simmetrica se se tutti gli elementi sotto alla diagonale principale sono zero.

As esempio

```python
M = [[3, 0, 1], [0, 0, 5], [0, 0, 6]] 
print(trisup(M))
```
