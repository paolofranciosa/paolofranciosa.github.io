## Implementazione

```python
import math

def fattorizza_in_lista(n):
    '''dato un naturale n, restituisce la fattorizzazione in fattori primi.
I fattori sono restituiti in una lista non decrescente'''
    fatt = []
    div = 2
    max_div = int(math.sqrt(n)) + 1
    while n > 1 and div <= max_div:
        if n % div == 0:
            fatt.append(div)
            n //= div
            max_div = int(math.sqrt(n)) + 1
        else:
            div += 1
    if n > 1:
        fatt.append(n)
    return fatt
```

## Traccia

Sappiamo che un numero naturale può essere scritto come il prodotto di numeri primi in un solo modo, se trascuriamo l'ordine dei fattori.
Ad esempio:
$$
444 = 2 * 2 * 3 * 37
$$

La fattorizzazione di un numero naturale consiste nel determinare questi fattori.

Dato un numero $n$ vogliamo determinare la lista dei suoi fattori primi, dove ciascun fattore compare con la dovuta molteplicità.

Per esempio, la fattorizzazione di $n=444$ produce la lista

$$    
[2, 2, 3, 37]
$$

Mentre, partendo da $n = 19$, poiché $n$ è primo, otteniamo la lista

$$
[19]
$$
