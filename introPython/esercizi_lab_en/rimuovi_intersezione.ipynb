## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione.

```python
def rimuovi_intersezione(a, b):
    tmp = b[:]
    for x in a:
        if x in b:
            b.remove(x)
            
    for x in tmp:
        if x in a:
            a.remove(x)
```

## Traccia 

Scrivere una funzione `rimuovi_intersezione(a, b)` che:

-   si aspetti come argomenti due liste `a`, `b`;
-   modifichi direttamente le liste `a` e `b` rimuovendo gli elementi che sono presenti in entrambe le liste.
    Non è necessario restituire un valore dalla funzione.

Ad esempio, se `a = [1, 2, 3]` e `b = [3, 4, 5]`, dopo aver chiamato `rimuovi_intersezione(a, b)`, `a` dovrebbe essere `[1, 2]` e `b` `[4, 5]`.

```python
a = [1, 2, 3]
b = [3, 4, 5]
rimuovi_intersezione(a,b)
print(a,b)
```
