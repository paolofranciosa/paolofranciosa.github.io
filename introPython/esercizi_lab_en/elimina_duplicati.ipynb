## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

````python
def elimina_duplicati(lista):
    if not isinstance(lista, list):
        raise TypeError

    if len(lista) == 0:
        return []

    results = [lista[0]]

    for parola in lista:
        if parola < results[-1]:
            raise ValueError
        elif parola != results[-1]:
            results.append(parola)
    return results
````

## Traccia
