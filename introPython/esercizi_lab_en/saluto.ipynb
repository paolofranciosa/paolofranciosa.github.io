```python
#!/usr/bin/env python3
def saluto(nome, anni):
    stampa = "Ciao, sono " + nome + " e ho " + str(anni)
    if anni > 1 or anni == 0:
        stampa = stampa + " anni"
    else:
        stampa = stampa + " anno"
    return stampa
```
