## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def rimuoviripetizioni(stringa):
    if len(stringa) == 0:
        return ''
    
    stringa_nuova = stringa[0]
    for carattere in stringa:
        if not (carattere in stringa_nuova):
            stringa_nuova = stringa_nuova + carattere
            
    return stringa_nuova 
```

## Traccia

Scrivete una funzione `rimuoviripetizioni(s)` che prende una stringa `s` in input e restituisce una stringa uguale, ma priva di ripetizioni dello stesso carattere, anche in posizioni non adiacenti. Ad esempio

```python
print(rimuoviripetizioni('allietare'))
```
