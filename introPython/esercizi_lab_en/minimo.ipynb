## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def minimo(valori, start=0):
    if not isinstance(valori, list):
        raise TypeError
    if len(valori) == 0:
        raise ValueError
    if start < 0 or start >= len(valori):
        raise ValueError
    for x in valori:
        if not isinstance(x, (int, float)):
            raise TypeError
    pos_min = start
    for i in range(start, len(valori)):
        if valori[i] < valori[pos_min]:
            pos_min = i
    return pos_min
```

## Traccia

Scrivere una funzione `minimo(valori, start=0)` che si aspetti come argomento una lista di valori ed eventualmente un intero che specifica la posizione iniziale, e restituisca la posizione del valore minimo tra gli elementi della lista che compaiono a partire dalla posizione `start` fino alla fine della lista. 

La lista di valori sarà non vuota e `start` sarà un un numero compreso tra `0` e `len(start)-1` incluso. Il parametro `start` è opzionale, e vale se non viene specificato la funzione si dovrà comportare come se fosse `0`.

Ad esempio:

```python
print(  minimo( [2, -5.2, 6,3, -3, 2] )     )
print(  minimo( [2, -5.2, 6,3, -3, 2], 2)   )
```
