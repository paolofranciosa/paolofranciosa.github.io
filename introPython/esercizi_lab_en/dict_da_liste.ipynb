## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_da_liste(k_list, v_list):

    res = {}
    for i in range(len(k_list)):
        key = k_list[i]
        value = v_list[i]
        res[key] = value
    return res
```

## Traccia



Scrivi una funzione `dict_da_liste(k_list, v_list)` che:

-   accetti come argomenti due liste `k_list` e `v_list` di uguale lunghezza;
-   restituisca un `dict` le cui chiavi sono prese dagli elementi di `k_list` e i cui valori sono presi dagli elementi di `v_list`. L'elemento `v_list[0]` sarò associato alla chiave `k_list[0]`,l'elemento `v_list[1]` sarà associato alla chiave `k_list[1]` e così via.

Per esempio:

```python
print(dict_da_liste(["cat", "dog", "mouse"],
                             ["gatto", "cane", "topo"]))
```
