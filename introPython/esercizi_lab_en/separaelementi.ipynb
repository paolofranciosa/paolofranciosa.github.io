Data una lita di numeri, restituisce due liste contenenti rispettivamente gli elementi maggiori della media e gli elementi minori o uguali alla media

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def separaelementi(valori):

    if len(valori) == 0:
        raise ValueError

    for x in valori:
        if not isinstance(x, (int, float)):
            raise TypeError

    media = sum(valori) / len(valori)

    left = [x for x in valori if x > media]
    right = [x for x in valori if x <= media]

    return left, right
```
