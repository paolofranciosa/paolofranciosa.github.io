## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def change_mins_in_zero(valori):
    minimo = min(valori)
    for i in range(0,len(valori)):
        if valori[i] == minimo:
            valori[i] = 0
```

## Traccia

Write a function `change_mins_in_zero(values)`  that:

- expects a non-empty list of comparable values as an argument;
- directly modifies the `values` list by replacing the minimum values with zero.

If the minimum appears several times in the list, the function must replace all occurrences of the minimum value with zero.

#### Example

If `values = [3, 1, 4, 1, 5]`, after calling `change_mins_in_zero(values)`, `values` should become `[3, 0, 4, 0, 5]`.


```python
a = [3, 1, 4, 1, 5]
change_mins_in_zero(a)
print(a)
```
