## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_rimuovi(d, lista_chiavi):
    if not isinstance(d, dict):
        raise TypeError
    new_d = {}
    for k in d:
        if k not in lista_chiavi:
            new_d[k] = d[k]
    return new_d
 
```

## Traccia


Scrivete una funzione `dict_rimuovi(d, lista_chiavi)` che:

-   si aspetti come argomenti un dizionario e una lista di chiavi;
-   crei e restituisca un `dict`, che contiene solo le chiavi, insieme ai rispettivi valori, che non appartengono a lista_chiavi. 

Ad esempio:

```python
diz = {'dog': 'cane', 'cat': 'gatto', 'mouse': 'topo', 'fish': 'pesce'}
lista = ['dog', 'mouse']
diz_1 = dict_rimuovi(diz, lista)
print(diz_1)
```
