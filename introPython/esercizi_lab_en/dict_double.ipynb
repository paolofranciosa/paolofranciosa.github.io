## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_double(d):
    new_d = {}
    for k in d:
        new_d[k] = 2 * d[k]
    return new_d
```

## Traccia



Write a function `dict_double(d)` that:

-   takes as arguments a dictionary `d` with values of type `int` or `float`;
-   returns a new dictionary containing, for each key in `d`, an item with the same key and a value that is twice the value in `d`.

You may assume that alle values are numeric.

For example:

```python
d = {"Mario": 1000, "Gianna" : 8000, "Alberto": 4000}

print(dict_double(d))
```
