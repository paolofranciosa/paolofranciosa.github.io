## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
#!/usr/bin/env python3
import math

def soluzione_maggiore(a, b, c):
    if (a == 0):
        return -c / b

    d = b**2 - 4  * a * c 

    if d < 0:
        return None
    else:
        x1 = (-b+math.sqrt(d)) / (2 * a)
        x2 = (-b-math.sqrt(d)) / (2 * a)
        if x1 > x2:
            return x1
        else:
            return x2 
```

## Traccia 

Scrivere una funzione `soluzione_maggiore(a, b, c)` che:

- dati in input i coefficienti `a`, `b` e `c`;

- se l’equazione non ha soluzione allora la funzione deve restituire `None`.

Se `a` è zero dovrete trattare l’equazione come se fosse di primo grado. 
Consultare la documentazione della funzione `math.sqrt`

Esempi:
