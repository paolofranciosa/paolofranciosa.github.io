```python
def rimuovi_annidato(S, x):
    for elemento in S[:]:
        if type(elemento) == list:
            ricerca_annidata(elemento, x)
        else:
            if elemento == x:
                S.remove(elemento)
```

## Traccia 


Scrivere una funzione `rimuovi_annidato(L, x)` che:

- si aspetti come argomento una lista `L` e un elemento `x`;
- modifichi direttamente la lista `L`, rimuovendo tutte le occorrenze di `x` dalla lista a qualsiasi livello di annidamento;
- la funzione non deve ritornare nessun valore;

Per questo esercizio si può usare la ricorsione.

#### Esempi

```python
L=[[3, 5, 6], 15, 10, [5], [[10], [1, 0, 10]]]
rimuovi_annidato(L, 10)
print(L)
```
