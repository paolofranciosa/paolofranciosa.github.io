## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione.

```python
def prodotto_mat(M1, M2): 
    P = []
    for i in range(len(M1)):
        riga = []
        for j in range(len(M2[0])):
            s = 0
            for z in range(len(M1[0])):
                s = s + M1[i][z]*M2[z][j]
            riga.append(s)
        P.append(riga)
            
    return P 
```
