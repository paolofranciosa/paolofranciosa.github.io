## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

````python
def bilanciofinale(testo):
    
    if not isinstance(testo, str):
        raise TypeError
    data=testo.split()
    
    if len(data)%2 != 0:
        raise ValueError
    
    D = {}
    for i in range(0,len(data),2):
        nome = data[i]
        valore = data[i+1]
        if nome not in D:
            D[nome]=0
        D[nome]+=int(valore)
    
    return D

def bilanciofinalefile(nomefile):
    if not isinstance(nomefile,str):
        raise TypeError
    with open(nomefile) as f:
        x = f.read()
        return bilanciofinale(x)
````
