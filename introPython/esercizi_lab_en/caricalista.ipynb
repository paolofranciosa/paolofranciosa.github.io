## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

````python
def caricalista(nomefile):
    with open(nomefile) as f:
        x = f.read()
    nums = [int(n) for n in x.split()]
    return nums
````

## Traccia

Scrivere una funzione `caricalista(nomefile)` che riceva come argomento il nome di un file di testo, e che

**Importante**: i numeri nel file devono essere separati da spazi. Non devono essere necessariamente tutti sulla stessa riga, e tra loro ci possono essere arbitrariamente spazi, a capo e righe vuote. Ad esempio l'esempio precedente potrebbe essere scritto su file anche come
