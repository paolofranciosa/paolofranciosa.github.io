## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def scambia_caratteri(s, x, y):
    if not len(s) > x >= 0:
        return None    
    if not len(s) > y >= 0:
        return None
    t = x 
    x = min(x,y)
    y = max(t,y)
    r = s[:x] + s[y] + s[x+1:y] + s[x] + s[y+1:]
    return r
```

## Traccia

Scrivi una funzione `scambia_caratteri(s, x, y)` che prenda in input
- una stringa `s`
- due interi `x` e `y`

La funzione deve restituire una stringa uguale a `s` ma con i caratteri in posizione `x` e `y` scambiati. Se `x` o `y` non sono posizioni valide, la funzione deve restituire `None`.

Ad esempio

```python
print(scambia_caratteri('balletto', 1, 7))
```
