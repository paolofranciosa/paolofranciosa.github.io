# Punto di sella di una matrice
Data una matrice di numeri, restituisce le coordinate del punto di sella (se esiste), altrimenti restituisce `None`. Il punto di sella è massimo, senza duplicati, sulla colonna e minimo, senza duplicati, sulla riga

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

````python
def sella_mat(m):

    r = len(m)
    c = len(m[0])

    for i in range(r):
        for j in range(c):
            # Verifica la sella
            candidato = True
            value = m[i][j]
            # verifica colonna
            for i2 in range(r):
                if value >= m[i2][j] and i2 != i:
                    candidato = False
                    break
            if not candidato:
                continue
            # verifica riga
            for j2 in range(c):
                if value <= m[i][j2] and j2 != j:
                    candidato = False
                    break
            if candidato:
                return (i, j)

    return None
````

## Traccia



Un elemento `M[i][j]` di una matrice `M` si dice punto di sella se è
**strettamente** maggiore di tutti gli altri elementi sulla riga `i` ed è
**strettamente** minore di tutti gli altri elementi sulla colonna `j`.


Ad esempio, la matrice

```
9   4  16  10   4
2   5   7   6   2
8  12  13   9   1
```

ha un punto di sella alla riga 1 e colonna 2, il cui valore è 7. Questo poiché 7 è il
massimo della riga 1 ed è anche il minimo della colonna 2. Può essere facilmente dimostrato che una matrice contiene al massimo un solo punto di sella.

Scrivere una funzione `sella_mat(M)` che:

-   si aspetti come parametro una matrice `M`;
-   restituisca la coppia di indici `(i,  j)` del punto di sella, se
    esiste, e restituisca `None` se `M` non ha un punto di sella.
    

````python
M = [[9, 4, 16, 10, 4], [2, 5, 7, 6, 2], [8, 12, 13, 9, 1]]

sella_mat(M)
````

    
Ricordiamo che in Python noi rappresentiamo una matrice $R \times C$ come
una lista di lunghezza $R$, contenente $R$ liste tutte quante di
lunghezza $C$. Potete sempre assumere che l\'input di questo esercizio
sia una matrice ben formata, con $R \geq  1$ e $C \geq 1$, e che tutte
le liste che rappresentano le righe contengano $C$ valori numerici.
