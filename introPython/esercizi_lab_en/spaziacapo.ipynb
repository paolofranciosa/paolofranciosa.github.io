## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def spaziacapo(s):
    ris = ''
    for carattere in s:
        if carattere == ' ':
            ris = ris + '\n'
        else:
            ris = ris + carattere
    return ris
```

## Traccia

Scrivete una funzione `spaziacapo(s)` che crea e restituisce una copia di `s` in cui ogni spazio di `s` è sostituito con un simbolo a capo `\n`. Ad esempio

```python
print(spaziacapo("c'era una volta"))
```
