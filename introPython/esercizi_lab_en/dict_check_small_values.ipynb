Dato un dizionario con chiavi stringa e una lunghezza soglia, verifica che tutte le chiavi abbiano lunghezza minore o uguale alla soglia

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_check_small_values(d, soglia):
    for k in d:
        if d[k] >= soglia:
            return False
    return True
```

## Traccia



Write a function `dict_check_small_values(d, threshold)` that:

-   takes as arguments a dictionary `d` with values of type `int` or `float` and an number `threshold`;
-   returns `True` if all values are strictly smaller than `threshold`, and returns `False` otherwise.

You may assume that alle values are numeric.

For example:

```python
d = {"Mario": 1000, "Gianna" : 8000, "Alberto": 4000}

print(dict_check_small_values(d, 7000))
```
