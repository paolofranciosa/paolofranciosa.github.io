## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
#!/usr/bin/env python3


def formattazione_dhms(secondi):
    if secondi < 0:
        raise ValueError("mi aspetto un input non negativo")
    if secondi == 0:
        return "0 secondi."

    sec_in_min = 60
    sec_in_ora = sec_in_min * 60
    sec_in_giorno = sec_in_ora * 24

    giorni = secondi // sec_in_giorno
    secondi %= sec_in_giorno

    ore = secondi // sec_in_ora
    secondi %= sec_in_ora

    minuti = secondi // sec_in_min
    secondi %= sec_in_min

    finale = []

    if giorni == 1:
        finale.append("1 giorno")
    elif giorni > 1:
        finale.append(str(giorni) + " giorni")

    if ore == 1:
        finale.append("1 ora")
    elif ore > 1:
        finale.append(str(ore) + " ore")

    if minuti == 1:
        finale.append("1 minuto")
    elif minuti > 1:
        finale.append(str(minuti) + " minuti")

    if secondi == 1:
        finale.append("1 secondo")
    elif secondi > 1:
        finale.append(str(secondi) + " secondi")

    if len(finale) == 1:
        return finale[0] + '.'
    else:
        return ", ".join(finale[:-1]) + " e " + finale[-1] + "."
```

## Traccia

-   aspettarsi come argomento un numero intero `secondi`;
