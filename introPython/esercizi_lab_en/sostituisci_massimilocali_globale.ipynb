## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def sostituisci_massimilocali_globale(sequenza):
    if len(sequenza) == 0:
        return
    massimo = max(sequenza)
    massimi = []
    for i in range(1, len(sequenza) - 1):
        if sequenza[i - 1] < sequenza[i] and sequenza[i] > sequenza[i + 1]:
            massimi.append(i)
    for i in massimi:
        sequenza[i] = massimo
```

## Traccia



Scrivere una funzione `sostituisci_massimilocali_globale(valori)` che:

- si aspetti come argomento una lista `valori`;
- modifichi direttamente la lista `valori` sostituendo ogni massimo locale con il valore massimo globale presente nella lista.

Un elemento di una sequenza è un massimo locale se:

- non si trova ad un estremo della lista;
- è preceduto e seguito da elementi strettamente minori dell'elemento stesso.

Il valore massimo globale è il valore più alto presente in tutta la lista. Se la lista contiene meno di tre elementi, non ci sono massimi locali e la lista non va modificata.

### Esempio

Se `valori = [3, 6, 4, 5, 4, 2]` e il massimo globale è `6`, dopo aver chiamato `sostituisci_massimilocali_globale(valori)`, `valori` dovrebbe diventare `[3, 6, 4, 6, 4, 2]`.

```python
a = [3, 6, 4, 5, 4, 2]
sostituisci_massimilocali_globale(a)
print(a)
```
