## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def losanga_vuota(n):
    res = ''
    righe = 0
    while righe < n:
        colonne = 0
        while colonne < n:
            if (colonne + 1 == n-righe):
                res = res + '*'
            else:
                res = res + ' '
            colonne = colonne + 1
            
        colonne = 0
        while colonne < righe:
            if (colonne == righe-1):
                res = res + '*'
            else:
                res = res + ' '  
            colonne = colonne + 1
            
        res = res + '\n'
        righe = righe + 1    
    
    righe = 1
    while righe < n:
        colonne = 0
        while colonne < n:
            if (colonne == righe):
                res = res + '*'
            else:
                res = res + ' '
            colonne = colonne + 1
            
        colonne = 0
        while colonne < n-righe-1:
            if (colonne == n-righe-2):
                res = res + '*'
            else:
                res = res + ' '                
            colonne = colonne + 1            
        res = res + '\n'
        righe = righe + 1
    
    return res
```
