## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def triangolo2(n):
    righe = 0
    res = ''
    while righe < n:
        colonne = 0
        while colonne < n:
            if (colonne + 1 >= n-righe):
                res = res + '*'
            else:
                res = res + ' '
            colonne = colonne + 1
        res = res + '\n'
        righe = righe + 1  
        
    return res
```
