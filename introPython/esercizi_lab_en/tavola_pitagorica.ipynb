## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
import math

def tavola_pitagorica(N):
    tavola = ''
    if N == 0:
        return tavola
    column_size = 2 + math.floor(2*math.log(N, 10))
    format_string = '{:' + str(column_size) + 'd}'    
    righe = 1
    while righe <= N:
        colonne = 1
        while colonne <= N:
            tavola = tavola + format_string.format(righe*colonne)
            colonne = colonne + 1
        tavola = tavola + '\n'
        righe = righe + 1
    return tavola   
```

## Traccia

Per andare a capo nella stringa usare il carattere speciale `\n`. Per convertire un numero `x` in una stringa di lunghezza `l` usare `'{:ld}'.format(x)`. Per ulteriori informazioni conslutare https://pyformat.info
