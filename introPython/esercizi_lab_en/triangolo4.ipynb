## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def triangolo4(n):
    righe = 0
    res = ''
    while righe < n:
        colonne = 0
        while colonne <= righe:
            res = res + str(colonne+1) + ' '
            colonne = colonne + 1
        res = res + '\n'
        righe = righe + 1
    return res

#bonus
def triangolo4variante(n):
    righe = 0
    while righe < n:
        colonne = 0
        while colonne <= righe:
            print('{:2d}'.format(colonne+ righe*(righe+1)//2), end=' ')
            colonne = colonne + 1
        print()
        righe = righe + 1
```
