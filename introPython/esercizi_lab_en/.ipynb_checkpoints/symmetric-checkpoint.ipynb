## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione.

```python
def symmetric(M):
    for r in range(0, len(M)):
        for c in range(0, len(M[0])):
            if M[r][c] != M[c][r]:
                return False
    return True    
```

## Traccia


Write a function `symmetric(M)` that,

-   takes as argument a *SQUARED* matrix `M`, represented as a list of lists;
-   returns `True` if `M` is a symmetric matrix, `False` otherwise.

A matrix is symmetric if and only if `M[i][j] == M[j][i]` for all `i, j`.

For example:

```python
M = [[3, 2, 1], [2, 0, 5], [3, 5, 6]] 
print(symmetric(M))
```
