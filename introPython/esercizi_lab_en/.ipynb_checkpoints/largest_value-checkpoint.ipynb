## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def largest_value(val_1, val_2):
    return max(abs(val_1), abs(val_2))
```

## Traccia

Write a function `largest_value(v_1, v_2)` that:

-   takes two numbers `v_1` `v_2` as parameters;
-   returns the maximum among the absolute values of `v_1` and `v_2`.


```python
print(largest_value(-7, 5))
print(largest_value(4, 8))
print(largest_value(-3, -5.7))
```
