## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def cumulated(valori):
    res = [valori[0]]
    for i in range(1, len(valori)):
        res.append(res[i-1] + valori[i])
    return res
```

## Traccia

Write a function `cumulated(values)`  that:

- expects a non-empty list `values` of numeric values as an argument;
- creates and returns a list having the same length as `values`, in which i-th element contains the sum of elements of `values` from 0-th to i-th (included).

#### Example


```python
a = [3, 1, 4, 1, 5]
print(cumulated(a))
```
