# Check whether all rows in a matrix are dense
Given a matrix and a natural k, checks whether each row in a matrix of number contains at least k non-zero entries

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def less_non_zero(r, k):
    count = 0
    for el in r:
        if el != 0:
            count += 1
    if count < k:
        return True
    else:
        return False

def mat_dense_rows(m, k):

    for r in m:
        if less_non_zero(r, k):
            return False
    return True
```

## Traccia



Write a function `mat_dense_rows(m, k)` that:

-   takes as argument a matrix `m` and an integer `k`;
-   returns `True` if each row in `m` contains at least `k` elements different from 0, returns `False` otherwise.

You may assume that `m` is a matrix represented by a list of lists of elements, and `m` contains at least one row and one column.

For example:

```python
m = [[1, 2, 5, 0], [0, 3, 0, 5], [3, 0, 0, 0]]
print(mat_dense_rows(m, 2))
print(mat_dense_rows(m, 1))
```
