# Count zero rows in a matrix
Restituisce il numero di righe in una matrice che contengono esclusivamente valori zero

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def good_mat(m):
    if not isinstance(m, list):
        return False
    for row in m:
        if not isinstance(row, list):
            return False
    if len(m) == 0:
        return False

    if len(m[0]) == 0:
        return False

    for row in m:
        if len(row) != len(m[0]):
            return False
    return True

def zeroes(r):
    for e in r:
        if e != 0:
            return False
    return True

def mat_count_zero_rows(m):

    if not good_mat(m):
        raise TypeError
    counter = 0
    for row in m:
        if zeroes(row):
            counter += 1
    return counter
```

## Traccia



Write a function `mat_count_zero_rows(m)` that:

-   takes as argument a matrix `m`, represented as a list of lists;
-   returns the number of rows in `m` whose elements are all zeroes.

We represent an $r \times c$ matrix in Python as a list with length $r$, where each element is a list of length $c$. You may assume that the argument is well formed, with $r \geq  1$, $c \geq 1$, and all the lists representing rows contain `c` numeric elements.


```python
m = [[0, 0, 0], [0, 1, 0], [2, 1, 1], [0, 0, 0]]
print(mat_count_zero_rows(m))
```
