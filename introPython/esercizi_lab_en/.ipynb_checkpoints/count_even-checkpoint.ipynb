## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def count_even(L):
    count = 0
    for x in L:
        if x % 2 == 0:
            count += 1
    return count
```

## Traccia


Write a function `count_even(L)` that,

-   takes a list `L` as parameter, in which each element is an `int` value;
-   returns the number of even values in `L`.
You can assume that `L` is a list of `int` numbers.


```python
a = [2, -1, 0, 4, 8]
print(count_even(a))
```
