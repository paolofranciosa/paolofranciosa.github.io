## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_union(d1, d2):
    unione = {}
    for k in d1:
        unione[k] = d1[k]
    for k in d2:
        if k in d1:
            unione[k] = unione[k] + d2[k]
        else:
            unione[k] = d2[k]
    return unione
```

## Traccia



Write a function `dict_union(d1, d2)` that:

-   takes two dictionaries `d1` e `d2` as arguments, with numeric values;
-   returns a `dict` where keys are taken from keys in `d1` and keys in `d2`. If a key is only in  `d1` or only in `d2` the associated value will be set in the returned dictionary. If a key is both in `d1` and in `d2` the value in teh returned dictionary will be the sum of the two values in `d1` and `d2`.

For example:

```python
d1 = {"January": 10000, "February" : 8000, "March": 4000}
d2 = {"January": 3000, "April" : 2000, "May": 1000}

print(dict_union(d1, d2))
```
