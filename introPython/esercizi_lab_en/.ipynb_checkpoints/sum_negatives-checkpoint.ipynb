## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def sum_negatives(L):
    total = 0
    for x in L:
        if x < 0:
            total += x
    return total
```

## Traccia


Write a function `sum_negatives(L)` that,

-   takes a list `L` as parameter, in which each element is a number (`int` or `float`);
-   returns the sum of all negative numbers in the list
    `L`.
You can assume that `L` is a list of numbers (each element is either `int` or `float`).


```python
a = [3, -1, 0, -4, 7]
print(sum_negatives(a))
```
