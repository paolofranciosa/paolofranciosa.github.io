## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def list_difference(a, b):

    if not isinstance(a, list) or not isinstance(b, list):
        raise TypeError

    res = []
    for x in a:
        if x not in b:
            res.append(x)
    return res
```

## Traccia

Write a function `list_difference(a, b)` that:

-   takes two lists `a`, `b`;

-   returns a new list containing all item in `a` that are not in `b`.

You can **assume** that `a` has no duplicates and `b` has no duplicates.

Elements in the result must appear in the same order as in `a`.


```python
x = [7, 5, 0, -1]
y = [-1, 6, 5]
print(list_difference(x, y))
```
