## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_stesse_chiavi(d1, d2):
    if not isinstance(d1, dict):
        raise TypeError
    if not isinstance(d2, dict):
        raise TypeError
    if len(d1) != len(d2):
        return False
    for k in d1:
        if k not in d2:
            return False
    return True
 
```

## Traccia


Scrivete una funzione `dict_stesse_chiavi(d1, d2)` che:

-   si aspetti come argomenti due dizionari;
-   restituisca `True` se i due dizionari contengono esattamente le stesse chiavi, `False` altrimenti. 

Ad esempio:

```python
diz_a = {'xx': 22, 'yy': 33, 'zz': 22}
diz_b = {'xx': 18, 'yy': 45, 'zz': 22}
print(dict_stesse_chiavi(diz_a, diz_b))
```
