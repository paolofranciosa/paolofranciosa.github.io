## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def somma_e_modifica_prima_lista(a, b):
    if len(a) != len(b):
        raise ValueError

    for i in range(len(a)):
        a[i] = a[i] + b[i]
```

## Traccia



Scrivere una funzione `somma_e_modifica_prima_lista(a, b)` che:

- si aspetti come argomenti due liste `a`, `b`, in cui ciascun elemento è un numero `int`;
- sollevi `ValueError` se le due liste non sono della stessa lunghezza;
- modifichi direttamente la lista `a` sommando a ciascun elemento i-esimo di `a` l'elemento i-esimo di `b`.

Dopo l'esecuzione, la lista `a` deve contenere la somma degli elementi omologhi di `a` e `b`.

#### Esempio

Se `a = [1, 2, 3]` e `b = [4, 5, 6]`, dopo aver chiamato `somma_e_modifica_prima_lista(a, b)`, `a` dovrebbe diventare `[5, 7, 9]`.

```python
a = [1,2,3]
b = [4,5,6]
somma_e_modifica_prima_lista(a, [4,5,6])
print(a,b)
```
