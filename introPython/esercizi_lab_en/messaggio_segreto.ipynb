````python
def messaggio_segreto(nomefile):
    if not isinstance(nomefile, str):
        raise TypeError
    with open(nomefile) as f:
        righe=f.readlines()
    parola=[]
    for riga in righe:
        ch=[c for c in riga if c.isupper()]
        if len(ch)==1:
            parola.append(ch[0])
    return "".join(parola)

import random
def produci_testo(r,c):
    a=ord('a')
    z=ord('z')
    righe=[]
    for i in range(r):
       row = [ chr(random.randint(a,z)) for _ in range(c) ]
       row = "".join(row)
       righe.append(row)
    return "\n".join(righe)    
````
