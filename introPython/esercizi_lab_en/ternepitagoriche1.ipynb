# Terne pitagoriche
Restituisce il numero di terne pitagoriche (contando anche terne proporzionali) formate da interi minori di $N$. Usa cicli while, complessità $N^3$, migliorabile a $N^2$

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def ternepitagoriche1(N):
    if not isinstance(N, int):
        raise TypeError

    counter = 0
    a = 1
    while a <= N:
        b = a + 1
        while b <= N:
            c = b + 1
            while c <= N:
                if a**2 + b**2 == c**2:
                    counter = counter + 1
                c += 1
            b += 1
        a += 1
    return counter
```
