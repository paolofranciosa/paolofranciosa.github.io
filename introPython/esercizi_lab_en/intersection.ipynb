## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def intersection(a, b):
    res = []

    for x in a:
        if x in b:
            res.append(x)
    return sorted(res)
```

## Traccia 

Write a function `intersection(a, b)` that:

-   takes as arguments two lists `a`, `b`;
-   returns a  **sorted** list representing the intersection between the two
    lists, that is, the elements that belong to both list
    `a` and the list `b`. It can be assumed that
    `a` and `b` contain no
    duplicates.
   
To get a sorted copy of a list you can call function `sorted(...)`
    
For example:
