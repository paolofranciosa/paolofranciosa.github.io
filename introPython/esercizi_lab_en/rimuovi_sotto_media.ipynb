## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione.

```python
def rimuovi_sotto_media(L):
    if len(L) ==0:
        raise ValueError
    
    m = sum(L)/len(L)
    tmp = L[:]
    for x in tmp:
        if x < m:
            L.remove(x)
```

## Traccia


Scrivere una funzione `rimuovi_sotto_media(L)` che:

- si aspetti come argomento una lista `L`;
- sollevi un errore `ValueError` se l'argomento ricevuto è una lista vuota;
- modifichi direttamente la lista `L` rimuovendo tutti gli elementi che sono minori della media aritmetica dei numeri contenuti nella lista `L`.

Si può assumere che la lista `L` contenga esclusivamente elementi numerici (`int` oppure `float`). La funzione non deve restituire un valore, ma deve modificare la lista `L`.

#### Esempio

Supponiamo che `L = [1, 2, 3, 4, 5]`. Dopo aver chiamato `rimuovi_sotto_media(L)`, poiché la media di `L` è 3, la lista modificata sarà `[3, 4, 5]`, rimuovendo quindi `1` e `2` che sono minori della media.


```python
L = [1, 2, 3, 4, 5]
rimuovi_sotto_media(L)
print(L)
```
