## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def moltiplicacaratteri(stringa, c, x):
    stringa_nuova = ''
    for carattere in stringa:
        if carattere == c:
            stringa_nuova = stringa_nuova + carattere*x
        else:
            stringa_nuova = stringa_nuova + carattere
            
    return stringa_nuova
```

## Traccia

Scrivete una funzione `moltiplicacaratteri(s, c, x)` che crea e restituisce una copia di `s` in cui ogni occorrenza di `c` è sostituita con `x` occorrenze consecutive di `c`. Ad esempio

```python
print(moltiplicacaratteri('abcb','b',3))
```
