## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

````python
def caricamatrice(nomefile):
    if not isinstance(nomefile, str):
        raise TypeError
    with open(nomefile) as f:
        x = f.read()
    nums = [int(n) for n in x.split()]
    if len(nums)<2: 
        raise ValueError
    R,C = nums[0],nums[1]
    if R<=0 or C<=0 or len(nums)!=2+R*C:
        raise ValueError
    
    M = [ [0]*C for _ in range(R) ]

    for r in range(R):
        for c in range(C):
            M[r][c] = nums[2+C*r+c]
    return M
````

## Traccia

Scrivere una funzione `caricamatrice(nomefile)` che riceva come argomento il nome di un file di testo, e che 
carichi da quel testo il contenuto di una matrice di interi. La funzione deve interpretare il contenuto del file come una serie di numeri **interi**.  I primi due numeri indicano il numero di righe e di colonne della matrice, e il resto dei numeri sono il contenuto delle celle. Ad esempio il file `caricamatrice3x5.txt` contenente 
```
3 5

2 0 1 -4 5 
6 7 10 -9 1 
2 4 5 5 0
```
e rappresenta la matrice
$$
\begin{bmatrix}
2 & 0 & 1 &- 4 & 5 \\ 
6 & 7 & 10 & -9 & 1 \\
2 & 4 & 5 & 5  & 0
\end{bmatrix}
$$
Quindi, per esempio,

````python
caricamatrice('caricamatrice3x5.txt')
````

**Importante**: i numeri nel file devono essere separati da spazi. Non devono essere necessariamente su righe diverse, e tra loro ci possono essere arbitrariamente spazi, a capo e righe vuote. Ad esempio l'esempio precedente potrebbe essere scritto su file anche come
```
3 5 2 0 1 -4 5 6 7 10 -9 1 2 4 5 5 0
```
Suggeriamo, una volta ottenuto il testo contenuto nel file in forma di stringa, di usare semplicemente il metodo `.split()` per estrarre i dati. 

Il formato **corretto** del file richiede che il file inizi con due numeri **interi positivi** $R$ e $C$, e prosegua con $R \times C$ numeri interi arbitrari. Nell'esempio precedente il numero di righe è $3$, il numero di colonne è $5$ e quindi ci devono essere esattamente altri $15$ numeri, nel file.

Il programma deve sollevare i seguenti errori:

  - `FileNotFoundError` se il file non esiste
  - `ValueError` se il file non è nel formato corretto
