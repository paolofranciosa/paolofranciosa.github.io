## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def piegature(spessore_finale):
    spessore = 0.01
    n_piegature = 0
    while spessore < spessore_finale:
        n_piegature = n_piegature + 1
        spessore = spessore * 2
        
    return n_piegature
```

## Traccia 

Scrivere una funzione `piegature(x)` che:

-   si aspetti come argomento un numero non-negativo `x` che rappresenta una lunghezza in millimetri;
-   restituisca quante volte un foglio dello spessore di 0.01 mm va piegato a metà per raggiungere (ed eventualmente superare) lo spessore di `x` mm;
-   se `x` è più piccolo o uguale allo spessore del foglio, restituisca 0.


Esempi:
- `piegature(0.02)` ritorna il valore `1`
- `piegature(0.03)` ritorna il valore `2`
- `piegature(500)` ritorna il valore `16`.
