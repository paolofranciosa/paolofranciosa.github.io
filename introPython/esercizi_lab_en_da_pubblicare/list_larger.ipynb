## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def list_larger(a, v):

    res = []
    for x in a:
        if x > v:
            res.append(x)
    return res
```

## Traccia

Write a function `list_larger(a, v)` that:

-   takes a lists `a` and a value `v`;

-   returns a new list containing all item in `a` that are strictly larger `v`.

You can **assume** that `a` contains elements that are comparable with `v`.

Elements in the result must appear in the same order as in `a`.


```python
x = [7, 5, 3, 8, -1]
k = 3
print(list_larger(x, k))
```
