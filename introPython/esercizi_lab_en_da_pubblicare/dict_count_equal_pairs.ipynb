# Count equal pairs in a dictionary
Counts how many items in a dictionary have value equal to the key

## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_count_equal_pairs(d):
    count = 0
    for k in d:
        if d[k] == k:
            count += 1
    return count
```

## Traccia



Write a function `dict_count_equal_pairs(d)` that:

-   takes as arguments a dictionary `d`;
-   returns the number of items in the dictionary in which the value associate dto the key is equal to the key itself.


For example:

```python
d = {"Mario": "Mario", "Gianna" : 8000, "Alberto": 4000, "John": "John"}

print(dict_count_equal_pairs(d))
```
