## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione

```python
def albero_numeri(n):
    res = ''
    righe = 0
    while righe < n:
        colonne = 0
        while colonne < n-righe-1 :
            res = res + ' '
            colonne = colonne + 1
            
        colonne = 0
        while colonne <= righe:
            res = res + str(colonne+1) + ' '
            colonne = colonne + 1
           
        res = res + '\n'
        righe = righe + 1
    
    return res
```

## Traccia 

Scrivere una funzione `albero_numeri(x)` che:

-   si aspetti come argomento un numero intero `x` compreso tra 0 e 10 (escluso);
-   restituisca una stringa che rappresenta un albero di numeri come nella figura sotto.

Per andare a capo nella stringa usare il carattere speciale `\n`.

Esempi:
- `albero(3)` ritorna il valore `'  1 \n 1 2 \n1 2 3 \n'`

La funzione deve essere salvata in un file che si chiama `albero_numeri.py`
