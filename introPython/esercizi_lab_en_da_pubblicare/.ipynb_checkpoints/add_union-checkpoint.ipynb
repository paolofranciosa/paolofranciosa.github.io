## Implementazione

La soluzione sarà disponibile in questo stesso file, dopo l'esercitazione.

```python
def add_union(a, b):
    for x in b:
        if x not in a:
            a.append(x)
```

## Traccia

Write a function `add_union(a, b)` that:

- takes two lists `a`, `b` as arguments;
- modifies list `a` adding all elements that appear in `b` and do not appear in `a`, so that `a` is the union of `a` and `b`.

You can assume that `a` and `b` do not contain duplicates. Anyway, some elemnts may appear both in `a` and in `b`. The function does not return any value, but leaves list `a` modified.

Elements must be added at the end of `a` in the same order n which they appear in `b`.

#### Example

If `a = [1, 2, 3]` and `b = [3, 4, 5]`, after calling `add_union(a, b)`, `a` should be `[1, 2, 3, 4, 5]`.


```python
a = [1,2,3]
b = [3,4,5]
add_union(a,b)
print(a,b)
```
