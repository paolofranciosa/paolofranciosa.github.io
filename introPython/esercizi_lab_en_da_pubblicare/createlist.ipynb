## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
import math

def createlist(a, b):
    if not isinstance(a, (int, float)) or not isinstance(b, int):
        raise TypeError
    if b < 0:
        raise ValueError
    l = []
    a = math.ceil(a)
    for i in range(b):
        l.append(a + i)
    return l
```

## Traccia

For example, the instruction `s = crealist1(14.3, 5)` must assign to `s` the list `[15, 16, 17, 18, 19]`.
