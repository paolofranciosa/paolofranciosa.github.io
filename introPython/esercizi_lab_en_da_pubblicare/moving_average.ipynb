## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def moving_average(seq, k):
    if not isinstance(seq, list):
        raise TypeError
    for x in seq:
        if not isinstance(x, (int, float)):
            raise TypeError

    if len(seq) < k or k < 1:
        raise ValueError

    medie = [0]
    for i in range(k):
        medie[0] += seq[i]

    for i in range(k, len(seq)):
        medie.append(medie[-1] + seq[i] - seq[i - k])
    for i in range(len(medie)):
        medie[i] /= k

    return medie
```

## Traccia

Write a function `moving_average(seq, k)` that:

-   takes as arguments a list of numbers `seq` and an integer `k`;
-   raises an exception `ValueError` if the list has zero length;
-   raises an exception `ValueError` if
    `0<k<=len(seq)` does not hold;
-   builds and returns the list of
     `len(seq)-k+1` [moving averages](https://en.wikipedia.org/wiki/Moving_average) of order
    `k`.

A moving average of order `k` is the arithmetic average of `k`
consecutive elements in the sequence. If the sequence has lenght `n` we can compute `n-k+1` moving averages: from the subsequence `[0:k]` to the subsequence `[n-k:n]` (left element included, right element not included). Note that the number of moving averages depends both on `n` and on `k`.

For example, if function `moving_average(seq, k)` is called with parameters `seq = [2, 4,  3, 6, 8, 9,  10, 12]` and `k = 4`, the
function must return `[3.75, 5.25,  6.5, 8.25,  9.75]`, that is the averages computed on the 5 subsequences `[2, 4, 3, 6]`, `[4, 3, 6, 8]`, `[3, 6, 8, 9]`, `[6, 8, 9, 10]`, `[8, 9, 10, 12]`.

```python
moving_average([2, 4,  3, 6, 8, 9,  10, 12],4)
```
