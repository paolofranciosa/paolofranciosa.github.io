```python
#!/usr/bin/env python3

def discount_price(prezzo, percentuale):
    if prezzo < 0 or percentuale < 0:
        return None
    return round(prezzo - prezzo/100 * percentuale, 1)
```

## Traccia 

Write a function `discount_price(p, s)` that:

- takes as arguments a price `p` and a discount percentage `s`, which are two numbers.
- returns the discounted price, rounded to the first decimal place;
- if `p` or `s` are negative, return `None`.

See the documentation for the `round` function

Examples:
- `discount_price(100, 10)` returns `90`.
- `discount_price(-100, 10)` returns the value `None`.

The function must be saved in a file called `discount_price.py`.
