## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

````python
def scalar_product(a, b):

    if not isinstance(a, list) or not isinstance(b, list):
        raise TypeError
    if len(a) != len(b):
        raise ValueError
    for i in range(len(a)):
        if not isinstance(a[i], (int, float)) or not isinstance(b[i], (int, float)):
            raise TypeError

    acc = 0
    for i in range(len(a)):
        acc += a[i] * b[i]

    return acc
````

## Traccia

The scalar product between two vectors $\langle a_0, a_1,\ldots, a_{n-1}
   \rangle$ and $\langle b_0,b_1,\ldots, b_{n-1} \rangle$ is defined as

```{=latex}
\begin{equation*}
      \sum_{i=0}^{n-1} a_i b_i
\end{equation*}
```
We represent vectors as lists of numbers.

Write a function `scalar_product` that:

-   takes two non-empty lists `a`, `b` with the same length as parameters, where each element is a number (`int` or `float`);

-   returns the scalar product of the two lists.

The scalar product between two lists is therefore the sum of the products of the
homologous elements. For example, the scalar product between the vectors `[1, 4, 3]` and
`[5, 5, 1]` is $1 \times 5 + 4 \times 5 + 3 \times 1 = 28$. The scalar product of two vectors is a numerical value.

You can **assume** that the lists are not empty, that they contain only numbers
and that they have the same length.
