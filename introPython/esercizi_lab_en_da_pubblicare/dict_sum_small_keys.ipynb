## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def dict_sum_small_keys(d, k_threshold):
    somma = 0
    for k in d:
        if k < k_threshold:
            somma += d[k]
    return somma
```

## Traccia



Write a function `dict_sum_small_keys(d, threshold)` that:

-   takes as argument a dictionary `d` with numeric values and a key `threshold`;
-   returns the sum of all values associated to keys smaller than `threshold`.

You may assume that all values in the dictionary are numeric.
If the dictionary is empty or it does not contain any key smaller than `threshold`, the functin must return 0.

For example:

```python
d = {"Mario": 1000, "Gianna" : 8000, "Alberto": 4000}

print(dict_sum_small_keys(d, "Lucia"))
```

Because keys `Alberto` and `Gianna` are smaller than `Lucia` in lexicografic order, while key `Mario` is larger.
