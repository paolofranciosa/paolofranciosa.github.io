## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def good_mat(m):
    if not isinstance(m, list):
        return False
    for row in m:
        if not isinstance(row, list):
            return False
    if len(m) == 0:
        return False

    if len(m[0]) == 0:
        return False

    for row in m:
        if len(row) != len(m[0]):
            return False
    return True

def dominates(a, b):
    for i in range(len(a)):
        if a[i] <= b[i]:
            return False
    return True

def mat_dominant_rows(m):

    if not good_mat(m):
        raise TypeError
    res = []
    for i in range(len(m)):
        domin = False
        for j in range(len(m)):
            if dominates(m[i], m[j]):
                domin = True
        if domin:
            res.append(i)
    return res
```

## Traccia



Write a function `mat_dominant_rows(m)` that:

-   takes as argument a matrix `m` of numbers, represented as a list of lists;
-   returns a list containing the positions of all rows of `m` that dominate at least one row in `m`.

Given two lists `a` and `b` with same length, we say `a` dominates `b` if and only if each element of `a` is strictly larger than the corresponding element of `b`. For example, [1, 3, -4, 6] dominates [0, 1, -9, 0] because 1 > 0, 3 > 1, -4 > -9 and 6 > 0. 

We represent an $r \times c$ matrix in Python as a list with length $r$, where each element is a list of length $c$. You may assume that the argument is well formed, with $r \geq  1$, $c \geq 1$, and all the lists representing rows contain `c` numeric elements.

In the following example:

- row in position 1 dominates row in position 4,
- row in position 2 dominates both row in position 0, row in position 3 and row in position 4

thus the result is `[1, 2]`

```python
m = [[0, 0, 0], [0, 1, 0], [2, 1, 1], [0, 0, 0], [-1, 0, -1]]
print(mat_dominant_rows(m))
```
