## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
import math

def crealista2(a, b, passo=1):
    if not isinstance(a, (int, float)) or not isinstance(
            b, int) or not isinstance(a, (int, float)):
        raise TypeError
    if b < 0:
        raise ValueError
    l = []
    a = math.ceil(a)
    for i in range(b):
        l.append(a + i * passo)
    return l
```

## Traccia

Scrivere una funzione `crealista2(a,  b, passo=1)`, con un

`a` con passo uguale al terzo argomento. Ad esempio,
l'istruzione 

`s = crealista2(14.6, 5, 3)` 

deve assegnare a `s` la lista `[15, 18, 21, 24, 27]`

Il terzo parametro può essere un intero qualunque. Ad esempio `crealista2(14.6, 5, -3)` deve restituire la lista `[15, 12, 9, 6, 3]`.
