## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def word_count(testo):

    if not isinstance(testo, str):
        raise TypeError

    counter = 0
    inword = False
    for c in testo:
        wordchar = c.isalpha() or c.isdigit()
        if wordchar:
            if not inword:
                counter += 1
            inword = True
        else:
            inword = False

    return counter
```

## Traccia

Write a function `word_count(text)` that returns the number of words in string `text`.
E.g.:


```python
print(word_count("Python4Dummy? Is the worst book on Python."))
```


Note that `Python4Dummy` is a single word. We consider as a word any sequence of uppercase or lowercase letters and digits. Spaces and punctuation are considered as separators between words.

You may use methods

-   `str.isalpha()` that returns `True` when a string is not empty and contains only letters;
-   `str.isdigit()` that returns `True` when a string is not empty and contains only digits.

In oparticular, you may use these methods to check wheter a single character is a letter or a digit.

For example:


```python
print('B'.isalpha())
print("3".isdigit())
print("a,b".isalpha())
print("Paul".isalpha())
print("342".isdigit())
print("Paul!".isalpha())
print("3 42".isdigit())
```
