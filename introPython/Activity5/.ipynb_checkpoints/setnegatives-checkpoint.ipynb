## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def setnegatives(valori):
    if not isinstance(valori, list):
        raise TypeError
    for x in valori:
        if not isinstance(x, (int, float)):
            raise TypeError
    for i in range(len(valori)):
        if valori[i] < 0:
            valori[i] = 0
```

## Traccia


Write a function `setnegatives(values)` that:

-   expects as an argument a list of numeric values (`int` or `float`)
-   edits the received list, setting all negative items to zero.

The function must return nothing, but it must modify the list passed as an argument. In particular, it is possible to write this function without using the `return` statement. Look in the next example how the function should behave


```python
L = [1.2, -0.4, 5, -2, 12 ]

setnegatives(L)

print(L)
```
