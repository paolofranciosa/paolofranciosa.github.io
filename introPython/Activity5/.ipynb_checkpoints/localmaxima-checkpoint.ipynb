## Implementazione

In questa sezione ci mettiamo la soluzione dell'esercizio, nel senso dell'implementazione. Questa deve essere messa prima della traccia, così la traccia potrà usare l'implementazione per far girare gli esempi.

```python
def localmaxima(sequenza):
    if not isinstance(sequenza, list):
        raise TypeError
    massimi = []
    for i in range(1, len(sequenza) - 1):
        if sequenza[i - 1] < sequenza[i] and sequenza[i] > sequenza[i + 1]:
            massimi.append(sequenza[i])
    return massimi
```

## Traccia



Write a function `localmaxima(seq)` that:

-   takes as argument a list `seq`;
-   creates and returns a list containing the local maxima in seq.

An element of a sequence is a local maximum if:

- it is not at an extreme of the list;
- it is preceded and followed by elements strictly less than the element
    itself.
   
A list containing fewer than three elements contains no local maximum, so an empty list must be returned in this case.
